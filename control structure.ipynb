### Create a simple Calculator

```python
# Simple Calculator
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
operation = input("Choose operation (+, -, *, /): ")

if operation == "+":
    print(f"Result: {num1 + num2}")
elif operation == "-":
    print(f"Result: {num1 - num2}")
elif operation == "*":
    print(f"Result: {num1 * num2}")
elif operation == "/":
    if num2 != 0:
        print(f"Result: {num1 / num2}")
    else:
        print("Error: Division by zero is not allowed.")
else:
    print("Invalid operation.")
```

### Find max and min in list

```python
# Find Max and Min in a List
def find_max_min(numbers):
    return max(numbers), min(numbers)

nums = list(map(int, input("Enter numbers separated by spaces: ").split()))
maximum, minimum = find_max_min(nums)
print(f"Maximum: {maximum}, Minimum: {minimum}")
```

### Find Prime Number

def is_prime(n):

    if n <= 1:  # Numbers less than or equal to 1 are not prime
    
        return False
    for i in range(2, n):  # Check divisibility from 2 to n-1
        if n % i == 0:  # If n is divisible by any number, it's not prime
            return False
    return True  # If no divisors were found, it's prime

# Get user input
num = int(input("Enter a number: "))

# Check and print the result
if is_prime(num):
    print(f"{num} is a prime number.")
else:
    print(f"{num} is not a prime number.")

### Factorial of a Number

```python
def factorial(n):
    # Base case: if n is 0 or 1, return 1
    if n == 0 or n == 1:
        return 1
    # Recursive case: n * factorial of (n - 1)
    else:
        return n * factorial(n - 1)

# Get input from user
num = int(input("Enter a number: "))

# Call the function and print the result
print(f"The factorial of {num} is {factorial(num)}.")
```

### Sum of digits

```python
# Function to calculate the sum of digits
def sum_of_digits(n):
    return sum(int(digit) for digit in str(abs(n))) 

# Input from the user
num = int(input("Enter a number: "))

# Print the sum of digits
print("Sum of digits:", sum_of_digits(num))
```

### Convert Decimal to Binary

```python
# Function to convert decimal to binary
def decimal_to_binary(n):
    return bin(n)[2:]  

# Input from the user
num = int(input("Enter a decimal number: "))

# Convert and print the binary representation
print("Binary representation:", decimal_to_binary(num))
```

### Greatest Common Divisor

```python
# Function to compute GCD using Euclid's Algorithm
def gcd(a, b):
    while b:
        a, b = b, a % b  # Swap and take remainder until b becomes 0
    return a

# Input from the user
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

# Compute and print the GCD
print("GCD:", gcd(num1, num2))
```

### Second Largest Number in a List

```python
# Function to find the second largest number
def second_largest(numbers):
    unique_numbers = list(set(numbers))  # Remove duplicates
    if len(unique_numbers) < 2:
        return "No second largest number"  # Handle cases with less than two unique numbers
    unique_numbers.sort(reverse=True)  # Sort in descending order
    return unique_numbers[1]  # Return second largest number

# Input: list of numbers
numbers = list(map(int, input("Enter numbers separated by spaces: ").split()))

# Output the second largest number
print("Second largest number:", second_largest(numbers))
```

### Count the Number of Digits in a Number

```python
# Function to count digits in a number
def count_digits(n):
    return len(str(abs(n)))  # Convert number to string (absolute value) and count characters

# Input from the user
num = int(input("Enter a number: "))

# Output the count of digits
print("Number of digits:", count_digits(num))
```

### Number is a Power of Two

```python
# Function to check if a number is a power of two
def is_power_of_two(n):
    return n > 0 and (n & (n - 1)) == 0  # Bitwise check for powers of two

# Input from the user
num = int(input("Enter a number: "))

# Output result
print(f"{num} is a power of two: {is_power_of_two(num)}")
```
